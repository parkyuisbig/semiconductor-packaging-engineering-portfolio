## Thermodynamics

평균온도가 아니라 vacuum 완료 시점의 온도, edge/center zone range, heating rate, overshoot와 gel-time proxy를 사용한다. Temperature는 점도를 낮추는 동시에 cure를 가속하므로 선형 단일효과로 가정하지 않는다.

## Surface roughness / interface

Ra 단독이 아니라 Rz, texture direction, contact angle, surface cleanliness와 coupon shear strength를 함께 본다. Roughness는 실접촉면적·마찰·열접촉저항·wetting을 동시에 바꿀 수 있으므로 최적 window를 검증한다.
